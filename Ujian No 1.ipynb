```python
import pandas as pd

# Lakukan load data dari file “Ujian_Komprehensif.csv” ke dalam Pandas DataFrame, beri nama df_sp2020; 
df_sp2020 = pd.read_csv('Ujian_Komprehensif.csv', delimiter=';')

# Tampilkan 10 baris pertama dari DataFrame yang telah terbentuk; 
print(df_sp2020.head(10))
```

```python

# Buatlah sebuah data frame baru dengan nama df_sp2020 dan pilih semua kolom kecuali kolom Total dan semua baris kecuali baris Total; 
df_sp2020 = df_sp2020.loc[
    df_sp2020['Provinsi'] != 'TOTAL',
    df_sp2020.columns != 'Total'
].copy()

df_sp2020
```

```python
# Buatlah pivot table dengan nama df_pivot_milenial dari df_sp2020 dengan menerapkan aggregate function count, sum pada Generasi ‘Milenial’, dengan index Pulau. Lakukan penyesuaian tipe data dan cleansing agar dapat dilakukan kalkulasi; 

df_milenial = df_sp2020.loc[df_sp2020['Provinsi'] != 'TOTAL', ['Pulau', 'Milenial']].copy()
df_milenial['Milenial'] = pd.to_numeric(
	df_milenial['Milenial'].astype(str).str.replace('.', '', regex=False),
	errors='coerce'
)
df_pivot_milenial = df_milenial.pivot_table(
	values='Milenial',
	index='Pulau',
	aggfunc=['count', 'sum']
)
df_pivot_milenial
```

```python
# Buatlah pivot table dengan nama df_pivot_post_Z dari df_sp2020 dengan menerapkan aggregate function count, sum pada Generasi ‘Post Generasi Z’, dengan index Pulau. Lakukan penyesuaian tipe data dan cleansing agar dapat dilakukan kalkulasi; 

df_post_Z = df_sp2020.loc[df_sp2020['Provinsi'] != 'TOTAL', ['Pulau', 'Post Generasi Z']].copy()
df_post_Z['Post Generasi Z'] = pd.to_numeric(
	df_post_Z['Post Generasi Z'].astype(str).str.replace('.', '', regex=False),
	errors='coerce'
)
df_pivot_post_Z = df_post_Z.pivot_table(
	values='Post Generasi Z',
	index='Pulau',
	aggfunc=['count', 'sum']
)
df_pivot_post_Z
```

```python
# Buatlah DataFrame gabungan dengan nama df_pivot_milenial_post_Z dengan nama kolom 'Count_Milenial', 'Sum_Milenial', 'Count_Post_Z', 'Sum_Post_Z'; 

df_pivot_milenial_post_Z = pd.concat([df_pivot_milenial, df_pivot_post_Z], axis=1)
df_pivot_milenial_post_Z
```

```python
# Simpan df_pivot_milenial_post_Z ke dalam file csv dengan delimiter "|" dan penamaan jawaban_bag1_[nip].csv (ganti [nip] dengan NIP 18-digit Anda). 

df_pivot_milenial_post_Z.to_csv(f'jawaban_bag1_198605252006021005.csv', sep='|', index=True)
```

#Bagian Ke 2

```python
# Salin DataFrame df_sp2020 ke dalam DataFrame baru bernama df_sp2020_2. Lakukan penyesuaian tipe data dan cleansing agar dapat dilakukan kalkulasi. Buatlah kolom baru dengan nama ‘TOTAL PENDUDUK’ pada df_sp2020_2. Kolom tersebut berisi jumlah penduduk untuk semua generasi per Provinsi; 

df_sp2020_2 = df_sp2020.copy()
kolom_generasi = [
    'Post Generasi Z', 'Generasi Z', 'Milenial',
    'Generasi X', 'Boomer', 'Pre Boomer'
]

df_sp2020_2[kolom_generasi] = df_sp2020_2[kolom_generasi].apply(
    lambda kolom: pd.to_numeric(
        kolom.astype(str).str.replace('.', '', regex=False),
        errors='coerce'
    )
)
df_sp2020_2['TOTAL PENDUDUK'] = df_sp2020_2[kolom_generasi].sum(axis=1)
df_sp2020_2['TOTAL PENDUDUK'].head(10)
```

```python
# Buatlah kolom baru pada df_sp2020 dengan nama ‘Kategori’. Kolom tersebut berisi kategori tingkat populasi dari kolom ‘TOTAL PENDUDUK’ dengan ketentuan sebagai berikut: 
# Kategori low adalah wilayah dengan populasi sampai dengan 3.000.000 penduduk; 
# Kategori mid adalah wilayah dengan populasi diatas 3.000.000 hingga 30.000.000 penduduk 
# Kategori high adalah wilayah dengan populasi di atas 30.000.0000 penduduk 
# (Bisa dengan membuat fungsi dan apply fungsi tersebut ke kolom ‘TOTAL PENDUDUK’); 

def kategori_populasi(total_penduduk):
    if total_penduduk <= 3000000:
        return 'low'
    elif 3000000 < total_penduduk <= 30000000:
        return 'mid'
    else:
        return 'high'

df_sp2020_2['Kategori'] = df_sp2020_2['TOTAL PENDUDUK'].apply(kategori_populasi)
df_sp2020_2['Kategori'].head(10)
```

```python
# Buat kolom baru dengan nama ‘kode_prop’ yang berisi angka kode propinsi yang diekstrak dari kolom ‘Provinsi’; 

df_sp2020_2['kode_prop'] = df_sp2020_2['Provinsi'].str.extract(r'(\d+)')[0].astype('Int64')
df_sp2020_2['kode_prop'].head(10)
```

```python
# Bersihkan angka didepan kolom ‘Provinsi’ sehingga hanya berisi nama provinsi saja tidak mengandung angka maupun titik; 

df_sp2020_2['Provinsi'] = df_sp2020_2['Provinsi'].str.replace(r'^\d+\.\s*', '', regex=True)
df_sp2020_2['Provinsi'].head(10)
```

```python
# Lakukan pivot (beri nama df_pivot_bag2) pada DataFrame dan urutkan ascending berdasarkan Provinsi sehingga terbentuk pivot seperti Gambar 2.; 
# hasilnya berupa kategori dan jumlah propinsi yang masuk ke dalam kategori tersebut.

df_sp2020_2['Kategori'] = df_sp2020_2['TOTAL PENDUDUK'].apply(kategori_populasi)

df_pivot_bag2 = (
    df_sp2020_2.loc[df_sp2020_2['Provinsi'] != 'TOTAL']
    .pivot_table(index='Kategori', values='Provinsi', aggfunc='count')
    .sort_values(by='Provinsi', ascending=True)
)
df_pivot_bag2
```

```python
# Simpan pivot hasil soal nomor 5 tersebut ke dalam file csv dengan delimiter |" dan format penamaan jawaban_bag2_[nip].csv (ganti [nip] dengan NIP 18-digit Anda). 

df_pivot_bag2.to_csv(f'jawaban_bag2_198605252006021005.csv', sep='|', index=True)
```

```python
# Buat dataframe baru dengan nama df_3 yang berisi kolom 'kode_prop', 'Provinsi', 'Pulau', 'TOTAL PENDUDUK', 'Kategori' dari dataframe df_sp2020_2; 

df_3 = df_sp2020_2[['kode_prop', 'Provinsi', 'Pulau', 'TOTAL PENDUDUK', 'Kategori']].copy()
df_3
```

```python
# Gunakan salah satu metode outlier detection untuk mendeteksi apakah kolom ‘TOTAL PENDUDUK’ masuk dalam kategori ‘Outlier’ atau ‘Not Outlier’ dan tampung hasilnya dalam kolom baru bernama ‘result_outlier’;

df_3['result_outlier'] = df_3['TOTAL PENDUDUK'].apply(lambda x: 'Outlier' if x > df_3['TOTAL PENDUDUK'].quantile(0.75) + 1.5 * (df_3['TOTAL PENDUDUK'].quantile(0.75) - df_3['TOTAL PENDUDUK'].quantile(0.25)) else 'Not Outlier')
df_3
```

```python
# Terapkan metode label encoding untuk mentransformasi isi kolom ‘result_outlier’ dengan ketentuan ‘Outlier’ menjadi 1 dan ‘Not oulier’ menjadi 0 dan tampung hasilnya dalam kolom bernama ‘is_outlier’; 

df_3['is_outlier'] = df_3['result_outlier'].map({'Outlier': 1, 'Not Outlier': 0})
df_3
```

```python
# Simpan DataFrame hasil soal nomor 3 tersebut ke dalam file csv dengan delimiter ";" dan format penamaan jawaban_bag3_[nip].csv (ganti [nip] dengan NIP 18-digit Anda); 

df_3.to_csv(f'jawaban_bag3_198605252006021005.csv', sep=';', index=False)
```
